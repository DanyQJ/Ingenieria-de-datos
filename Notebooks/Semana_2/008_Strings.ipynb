```python
ruta = "c:\tam\Projects\now"
print(ruta) # esto imprimira c:	am\Projectsnow debido a que \t y \n son caracteres de escape
ruta_escapada = "c:\\tam\\Projects\\now"
print(ruta_escapada) # esto imprimira c:\tam\Projects\now
ruta_raw = r"c:\tam\Projects\now"
print(ruta_raw) # esto imprimira c:\tam\Projects\now debido a que el prefijo r indica que la cadena es raw y no se interpretaran los caracteres de escape
```

## Interrogation

```python
leo = "Leonardo DiCaprio"
leo
```

```python
leo.index("DiCaprio") # el metodo index() devuelve el indice de la primera ocurrencia de la subcadena especificada, en este caso "DiCaprio" se encuentra a partir del indice 10
```

```python
leo.index("D") # el metodo index() devuelve el indice de la primera ocurrencia del caracter o subcadena especificada, en este caso devuelve 9 porque la letra D es la novena letra de la cadena
```

```python
leo.index("f") # si la subcadena no se encuentra en la cadena, el metodo index() lanza una excepcion ValueError
```

```python
leo.find("f") # el metodo find() devuelve el indice de la primera ocurrencia del caracter o subcadena especificada, pero a diferencia de index(), si la subcadena no se encuentra en la cadena, find() devuelve -1
```

```python
leo.startswith("Leo") # el metodo startswith() devuelve True si la cadena comienza con la subcadena especificada, en este caso devuelve True porque "Leonardo DiCaprio" comienza con "Leo"
```

```python
leo.endswith("rio") # el metodo endswith() devuelve True si la cadena termina con la subcadena especificada, en este caso devuelve True porque "Leonardo DiCaprio" termina con "rio"
```

## Content Type

```python
'abc123'.isalpha() # el metodo isalpha() devuelve True si todos los caracteres de la cadena son letras, en este caso devuelve False porque la cadena contiene numeros
```

```python
'abc123'.isalnum() # el metodo isalnum() devuelve True si todos los caracteres de la cadena son alfanumericos (letras o numeros) y hay al menos un caracter, en este caso devuelve True porque "abc123" es alfanumerica
```

```python
'lowercase'.islower() # el metodo islower() devuelve True si todas las letras de la cadena estan en minuscula y hay al menos una letra, en este caso devuelve True porque "lowercase" esta en minuscula
```

```python
'lowercase'.isupper() # el metodo isupper() devuelve True si todas las letras de la cadena estan en mayuscula y hay al menos una letra, en este caso devuelve False porque "lowercase" esta en minuscula    
```

```python
one = '1'
one.isnumeric() # el metodo isnumeric() devuelve True si todos los caracteres de la cadena son numeros y hay al menos un caracter, en este caso devuelve True porque "1" es un numero
```

## Format strings
1. Introduced in Python 3.6.
2. Prefixed by either a 'F' or 'f'.
3. Values inserted at runtime using replacement fields.

```python
f"Example {1}"
```

Insert variable into replacement field

```python
string_count = 5
frets_count = 24
f"Noam Chomsky has {string_count} strings and {frets_count} frets on his guitar" # las f-strings permiten incluir expresiones dentro de las llaves {} que seran evaluadas e insertadas en la cadena resultante
```

Insert expression into replacement field

```python
a = 10
b = 20
f"The sum of {a} and {b} is {a + b}" # las f-strings pueden incluir cualquier expresion valida de python dentro de las llaves {}, incluyendo indices de listas, operaciones matematicas, llamadas a funciones, etc
```

Index list in string replacement fields

```python
players = ["John", "Paul", "George", "Ringo"]
f"The Beatles were {players[0]}, {players[1]}, {players[2]} and {players[3]}" # las f-strings pueden incluir cualquier expresion valida de python dentro de las llaves {}, incluyendo indices de listas, operaciones matematicas, llamadas a funciones, etc
```

Padding a number

```python
lucky_number = 7
f"to pad the number {lucky_number} to 5 places:{lucky_number:5d}" # las f-strings permiten formatear los valores incluidos en las llaves {} utilizando especificadores de formato, en este caso :5d indica que el numero debe ser formateado como un entero con un ancho minimo de 5 caracteres, rellenando con espacios a la izquierda si es necesario
```

Setting padding value at runtime


```python
lucky_number = 7
padding = 5
f"to pad the number {lucky_number} to {padding} places:{lucky_number:{padding}d}" # las f-strings permiten utilizar variables para especificar el formato, en este caso {lucky_number:{padding}d} indica que el numero debe ser formateado como un entero con un ancho minimo de padding caracteres, rellenando con espacios a la izquierda si es necesario
```

Commas

```python
balance = 12348923938
print(f"Your balance is {balance:,}") # las f-strings permiten formatear los numeros con separadores de miles utilizando el especificador de formato :, en este caso el numero se formateara como 12,348,923,938
```

## Manipulate strings

Removing Whitespaces

```python
ship = " The Black Pearl "
print(ship.strip()) # el metodo strip() devuelve una nueva cadena con los espacios en blanco al principio y al final eliminados, en este caso devuelve "The Black Pearl"
print(ship.lstrip()) # el metodo lstrip() devuelve una nueva cadena con los espacios en blanco al principio eliminados, en este caso devuelve "The Black Pearl "
print(ship.rstrip()) # el metodo rstrip() devuelve una nueva cadena con los espacios en blanco al final eliminados, en este caso devuelve " The Black Pearl"
```

Spitting and Joining

```python
words_string = "uno,dos,tres,cuatro"
words_string
```

Split on comma

```python
words = words_string.split(",") # el metodo split() devuelve una lista de subcadenas obtenidas al dividir la cadena original utilizando el separador especificado, en este caso devuelve ["uno", "dos", "tres", "cuatro"]
words
```

Joining

```python
':'.join(words) # el metodo join() devuelve una nueva cadena que es la concatenacion de las cadenas en la lista especificada, separadas por la cadena que llama al metodo, en este caso devuelve "uno:dos:tres:cuatro"
```

Slip on newline

```python
multiline = "Sometimes we are given\na multiline document\nas a single string"
print(multiline) # el caracter \n representa un salto de linea, por lo que al imprimir la cadena se mostrarán las líneas separadas
print(multiline.splitlines()) # el metodo splitlines() devuelve una lista de las lineas en la cadena, dividiendo la cadena en cada salto de linea, en este caso devuelve ["Sometimes we are given", "a multiline document", "as a single string"]   
```
