Shared Sequence Operations:

```python
# Memberships
3 in [1,2,3,4,5] # esto lo utilizamos para probar si el numero se encuentra en esta lista
```

```python
12 in [1,2,3,4,5] # prueba si se encuentra el numero en la lista
```

```python
# Indexing
name = "Monteverdi"
name[0] # de esta manera podemos acceder a objetos en especifico en la secuencia
```

```python
name[4] # muestra otro objeto en especifico de la secuencia
```

```python
name[-1] # muestra el ultimo objeto de la secuencia
```

```python
name[-2]
```

```python
# Slicing
name = 'Palestrina'
name [2:5] # el primer numero representa el primer objeto en mostrar de esa secuencia
           # el segundo numero representa el ultimo objeto en excluir de la secuencia
```

```python
name[:5] # si dejamos sin numero el primer argumento, automaticamente asumira que es 0
```

```python
name[4:] # si dejamos sin numero el segundo argumento, automaticamente lo tomra has ta el valor del final
```

```python
name[-3:]
```

```python
name[:] # muestra toda la secuencia
```

```python
name[::2] # Esto se traduce en inicio:fin:paso ,donde el paso es el 'salto' que da el cursor entre palabras
```

```python
name[::-1] # esto truco es famoso ya que lo que hace es recorrer la lista hacia atras, dando pasos negativos
```

```python
# Interrogation
len(name) # Pregunta el cuantos objetos tiene mi secuencia
```

```python
min(name) # Esto encuentra el objeto mínimo de la secuencia
```

```python
max(name) # Esto encuentra el objeto maximo de la secuencia
```

```python
max(['Free',2,'b'])
```

```python
name.count('e') # Esto me permite contar cuantos objetos hay de 'e' en mi cadena
```

```python
name.index('i') # Va encontrar la primera en el indice por defecto
```

```python
# Math
"prefix" + "-" + "postfix"
```

```python
[1,2] + [3,4] # En lugar de sumar los items, crea una nueva lista con todos los items
```

```python
[0,2] * 4 # Crea una lista nueva con cuatro veces la lista original
```
