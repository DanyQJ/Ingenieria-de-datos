## Adding and Removing items

```python
composers = ['Vivaldi','Mozart']
composers
```

```python
composers.append('Beethoven') # con el metodo append() podemos agregar un elemento a una lista
composers
```

```python
composers.insert(0,'Bach') # con el metodo insert() podemos agregar un elemento a una lista en la posicion que queramos, en este caso lo agregamos en la posicion 0
composers
```

```python
composers.pop(0) # con el metodo pop() podemos eliminar un elemento de una lista, en este caso eliminamos el elemento en la posicion 0, que es 'Bach'
composers
```

```python
vips = ['Beyonce','Jay-Z','Kanye West']
vips
```

```python
vips.extend(composers) # con el metodo extend() podemos agregar los elementos de una lista a otra lista, en este caso agregamos los elementos de la lista composers a la lista vips
vips
```

```python
lists = [[]]*4 # esto crea una lista de 4 listas vacias, pero todas las listas vacias son la misma lista, por lo que si modificamos una de las listas vacias, se modifican todas las listas vacias
lists
```

```python
lists[-1].append(4) # esto agrega el numero 4 a la ultima lista vacia, pero como todas las listas vacias son la misma lista, se agrega el numero 4 a todas las listas vacias
lists
```

```python
lists = [[] for _ in range(4)] # esto crea una lista de 4 listas vacias, pero cada lista vacia es una lista diferente, por lo que si modificamos una de las listas vacias, no se modifican las otras listas vacias
lists[-1].append(4) # esto agrega el numero 4 a la ultima lista vacia, pero como cada lista vacia es una lista diferente, se agrega el numero 4 solo a la ultima lista vacia
lists
```

## Ordering

```python
name = "Beethoven"
```

```python
letters = list(name) # esto crea una lista a partir de una cadena, cada letra de la cadena se convierte en un elemento de la lista
letters
```

```python
letters.sort() # con el metodo sort() podemos ordenar una lista, en este caso ordenamos la lista letters, pero como las letras de la palabra Beethoven no estan en orden alfabetico, la lista letters no se ordena
letters
```

```python
letters.reverse() # con la funcion reverse() podemos invertir el orden de una lista, en este caso invertimos el orden de la lista letters, pero como las letras de la palabra Beethoven no estan en orden alfabetico, la lista letters no se invierte
letters
```
