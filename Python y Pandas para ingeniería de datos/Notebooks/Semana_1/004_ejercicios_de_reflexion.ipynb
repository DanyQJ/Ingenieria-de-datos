### Define una funcion que calcule el interes simple

```python
# Function to calculate simple interest 
def simple_interest(principal, rate, time):
    
    # Calculate simple interest
    interest = principal * rate * time
    
    # Return interest 
    return interest

principal = 5000
rate = 0.1 
time = 5

interest = simple_interest(principal, rate, time)
print(interest) # 2500.0
```

### Escribe una funcion que pruebe si la palabra es un palindromo

```python
def verificar_palindromo(palabra):
    return palabra == palabra[::-1]

verificar_palindromo('leche')
verificar_palindromo('ana')
```

### Escribe una funcion que encuentre el numero mas grande de 3 numeros dados

```python
def numero_mayor(a,b,c):
    if a > b and a > c:
        print(f'{a} es el numero más grande')
    elif b > a and b > c:
        print(f'{b} es el numero más grande')
    else:
        print(f'{c} es el numero mas grande')
```

```python
numero_mayor(10, 25, 5)
```
