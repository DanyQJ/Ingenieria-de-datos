# parameters

```python
def add_one(num):
    return num + 1
```

```python
add_one(5)
```

# parameters by order and by name

```python
def my_func(first, second, third):
    print(first)
    print(second)
    print(third)
```

```python
my_func(1,2,3) #llamando a los parametros por posicion
```

```python
my_func(third=3, first=1, second=2) #llamando a los parametros por nombre
```

# Setting default values

```python
def say_hello(name="World"): #parametro con valor por defecto
    print("Hello" + name)
```

```python
say_hello()
```
