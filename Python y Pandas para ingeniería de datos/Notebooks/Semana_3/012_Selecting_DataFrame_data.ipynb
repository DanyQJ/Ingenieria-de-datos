## Selecting DataFrame Data

```python
import pandas as pd
file_path = '../../data/USCG.Search.Rescue.Stats.csv'
df = pd.read_csv(file_path)
df.head()
```

### Boolean Mask

```python
# Creamos una máscara de booleanos con el mismo tamaño que el DataFrame, inicialmente con todos los valores en False
# _ significa que no necesitamos el valor de la variable, solo queremos repetir la operación len(df) veces
mask = [False for _ in range(len(df))] 
# Establecemos los valores de la máscara a True para las filas que queremos seleccionar (en este caso, las filas 3 a 6)
mask[3:7] = [True] * 4 
mask
```

```python
df[mask]
```

```python
# Otra forma de seleccionar filas usando la máscara es con loc, que es un indexador basado en etiquetas. En este caso, como la máscara es un array de booleanos, loc seleccionará las filas correspondientes a los valores True en la máscara.
df.loc[mask] 
```

### Creating mask using comparison operators

```python
mask = df.loc[:, 'Lives Lost After CG Notification'] < df.loc[:, 'Lives Lost Before CG Notification']
mask
```

```python
df[mask]
```

### Pandas Boolean operators

1. And: &
2. Or: |
3. Not: ~

```python
df.describe()
```

```python
# Seleccinamos las filas que cumplen con dos condiciones: que el número de casos sea menor a 60000 y que el número de salidas sea mayor a 4500. Para esto, usamos el operador & para combinar las dos condiciones y creamos una nueva máscara de booleanos.
mask = (df.loc[:, 'Cases']< 60000) & (df.loc[:, 'Sorties']> 4500)
df[mask]
```

### Creating new column

```python
# Creamos una nueva columna en el DataFrame llamada "Saved per Sortie" que contiene el número de vidas salvadas por cada salida. Para esto, dividimos la columna "Lives Saved" entre la columna "Sorties" y asignamos el resultado a la nueva columna.
df.loc[:,'Saved per Sortie'] = df.loc[:,'Lives Saved'] / df.loc[:,'Sorties']
```
